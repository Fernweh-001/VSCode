```python
n,s = 10,100
def fact(n):
    global s
    for i in range(1,n+1):
        s *= i
    return s
print(fact(n),s)
```

```python
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n-1)

fact(5)
```

```python
def rvs(s):
    if s == '':
        return s
    else:
        return rvs(s[1:])+s[0]
    
rvs('jfhjsd')
```

```python
def f(n):
    if n == 1 or n == 2:
        return 1
    else:
        return f(n-1) + f(n-2)
f(10)
```

```python
#汉诺塔
count = 0
def hannoi(n,src,dst,mid):
    global count
    if n == 1:
        print('{}:{}>-{}'.format(1,src,dst))
        count +=1
    else:
        hannoi(n-1,src,mid,dst)
        print('{}:{}>-{}'.format(n,src,dst))
        hannoi(n-1,mid,dst,src)
        count += 1
hannoi(3,'A','C','B')
print(count)
```
